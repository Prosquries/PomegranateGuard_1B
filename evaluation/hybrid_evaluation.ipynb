```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras.models import load_model
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
# Load Hybrid Model
model = load_model(
    "D:/PomegranateGuard/model/hybrid_model.h5",
    compile=False
)

print("Hybrid Model Loaded Successfully")

# Load Validation Dataset
X_val = np.load("D:/PomegranateGuard/dataset/X_val_autoencoder.npy")
y_val = np.load("D:/PomegranateGuard/dataset/y_val.npy")

print("Validation Data Loaded")
```

```python
# Predict
y_pred_probs = model.predict(X_val)
y_pred = np.argmax(y_pred_probs, axis=1)

# Accuracy
accuracy = accuracy_score(y_val, y_pred)
print("\nHybrid Validation Accuracy:", accuracy)

# Classification Report
report = classification_report(y_val, y_pred)
print("\nClassification Report:\n")
print(report)
```

```python
# Save report
with open("D:/PomegranateGuard/evaluation/hybrid_evaluation.txt", "w") as f:
    f.write("Validation Accuracy: " + str(accuracy) + "\n\n")
    f.write(report)

# Confusion Matrix
cm = confusion_matrix(y_val, y_pred)
```

```python
plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt="d")
plt.title("Hybrid Model Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.savefig("D:/PomegranateGuard/evaluation/hybrid_confusion_matrix.png")
plt.show()

print("Hybrid Evaluation Completed Successfully")
```
