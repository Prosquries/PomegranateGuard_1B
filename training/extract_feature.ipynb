```python
import numpy as np
from tensorflow.keras.models import load_model, Model

# load trained model
model = load_model("D:/PomegranateGuard/model/pomegranate_feature_extractor.h5")

feature_model = Model(
    inputs=model.input,
    outputs=model.layers[-2].output
)

# load dataset safely
X_train = np.load("D:/PomegranateGuard/dataset/X_train.npy", mmap_mode='r')
X_val   = np.load("D:/PomegranateGuard/dataset/X_val.npy", mmap_mode='r')
X_test  = np.load("D:/PomegranateGuard/dataset/X_test.npy", mmap_mode='r')

# extract features in batches
train_features = feature_model.predict(X_train, batch_size=16)
val_features   = feature_model.predict(X_val, batch_size=16)
test_features  = feature_model.predict(X_test, batch_size=16)

print("Train features:", train_features.shape)

# save features
np.save("D:/PomegranateGuard/features/train_features.npy", train_features)
np.save("D:/PomegranateGuard/features/val_features.npy", val_features)
np.save("D:/PomegranateGuard/features/test_features.npy", test_features)

print("Feature extraction completed")
```
