```python
import numpy as np

# Load dataset
X_train = np.load("D:/PomegranateGuard/dataset/X_train.npy")
X_val   = np.load("D:/PomegranateGuard/dataset/X_val.npy")

print("Original Shape:", X_train.shape)
```

```python
# Step 1: Convert to float32
X_train = X_train.astype('float32')
X_val   = X_val.astype('float32')

# Step 2: Normalize (IMPORTANT)
X_train = X_train / 255.0
X_val   = X_val / 255.0

print("Preprocessing Completed")
```

```python
# Save preprocessed data 
np.save("D:/PomegranateGuard/dataset/X_train_autoencoder.npy", X_train)
np.save("D:/PomegranateGuard/dataset/X_val_autoencoder.npy", X_val)

print("Autoencoder Preprocessed Data Saved")
```
